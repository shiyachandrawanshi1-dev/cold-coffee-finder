```python
from geopy.distance import geodesic
```

```python
user_location = (25.155,55.330)
spot1 = (25.213998,55.275438)
round(geodesic(user_location,spot1).km,2)
```

```python
def calculate_distance(user_location,spot_location):
    dis = round(geodesic(user_location,spot_location).km,2)
    return dis
```

```python
def get_distance(row):
    lat = row['lat']
    lng = row['lng']
    spot_location = (lat,lng)
    return calculate_distance(user_location,spot_location)



```

```python
df['distance'] = df.apply(get_distance,axis=1)
```

```python
from datetime import datetime
current_time = datetime.now().strftime("%H:%M")
```

```python
def is_open_spot(row):
    open_time = row['opening_time']
    close_time = row['closing_time']
    current_time = datetime.now().strftime("%H:%M")

    is_open = open_time <= current_time <= close_time
    if is_open:
        return 'Open'
    else:
        return 'Closed'
```

```python
df['is_open_spot'] = df.apply(is_open_spot,axis=1)
```
